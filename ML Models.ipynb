##En-Coding

###One-Hot encoding:

Description: This method creates new binary (0 or 1) columns for each unique category in a feature. If a data point belongs to a category, the corresponding column will have a '1', and all other category columns will have a '0'.

Use Cases: Best for nominal categorical variables (categories without a natural order) where the number of unique categories is not excessively large. It prevents the model from assuming any ordinal relationship between categories.

Example: If you have a 'Color' feature with categories 'Red', 'Green', 'Blue', it would be converted into three new columns: 'Color_Red', 'Color_Green', 'Color_Blue'.

###Label encoding:

Description: Assigns a unique integer to each category based on its alphabetical order or a predefined ordinal rank. For example, 'Red' might be 0, 'Green' 1, 'Blue' 2.

Use Cases: Suitable for ordinal categorical variables (categories with a natural, meaningful order). If used on nominal data, it can introduce an artificial order that might mislead the model.

Example: For a 'Size' feature with categories 'Small', 'Medium', 'Large', you could encode them as 0, 1, 2 respectively.

### Deriving Nominal and Ordinal Categories

Identifying whether a categorical variable is nominal (unordered) or ordinal (ordered) often requires domain knowledge about the features. While automated methods exist, they can be complex and may not always yield accurate results without context.

In this case, we use `df.select_dtypes(include='object')` to get all categorical columns. From these, `KitchenQual` is typically an ordinal feature (e.g., 'Ex' > 'Gd' > 'TA' > 'Fa' > 'Po'). The remaining categorical columns are considered nominal.

```python
all_categorical_cols = df.select_dtypes(include='object').columns.tolist()

ordinal_categorical_cols = ['KitchenQual']

nominal_categorical_cols = [col for col in all_categorical_cols if col not in ordinal_categorical_cols]

print("Nominal Categorical Columns (derived from DataFrame and domain understanding):")
print(nominal_categorical_cols)
print("\nOrdinal Categorical Columns (derived from DataFrame and domain understanding):")
print(ordinal_categorical_cols)
```
